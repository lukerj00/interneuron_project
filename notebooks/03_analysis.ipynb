```python
# load data

data = scipy.io.loadmat('../data/Data_cell_types_small_size.mat')
keys = list(data.keys())
data_keys = keys[3:7]
data_fit_list = [data[key] for key in data_keys]
contrasts = data[keys[-1]]
c_range = contrasts.flatten()/100

N_datapoints = np.array([len(cell_type) for cell_type in data_fit_list])
means = np.array([np.mean(cell_type, axis=0) for cell_type in data_fit_list])
covs = np.array([np.cov(np.array(cell_type).T) for cell_type in data_fit_list])
sems = np.array([np.sqrt(np.diag(cov) / N_datapoints[i]) for i, cov in enumerate(covs)])

# load params/fits from file

with open('../data/fits_cnd1_top200_250102_125616.pkl', 'rb') as f: 
    top_fits, top_params, top_likelihoods = pickle.load(f)

# global constants for analytics

TAU = np.diag([0.02,0.01,0.01,0.01]) # time consts (20ms E, 10ms I)
dt = 1e-4
tau_noise = 2e-2 # OU only (note 50ms 'slow noise' in Hennequin paper)
sigma_noise = 1e-2 # noise magnitude
variable = 'voltage' # variable to simulate (voltage or rate)
noise_type = 'white' # noise type (none, white or OU). OU only for rate

# select fit, contrast

k = 7 # fit index
contrast = 1 # contrast from c_range
fit = top_fits[k]
params = top_params[k]
c_ind = np.where(c_range == contrast)[0][0]
fit_c = fit[:,c_ind] # fixed point rates at fit, contrast
```

### hopf analysis

```python
# sim params

W, h1, h0 = params
T_sim = 500
variable = 'rate'
noise_type = 'none'
init_perturb_mag = 1e-3

# define external input range to check for bifurcation

h_ext_lower = 0.5*np.array([0, 0, 0, 1]) # 5465
h_ext_upper = 0.6*np.array([0, 0, 0, 1]) # 5495
points = 60
```

```python
# simulate linearised system before analysis

h_ext = np.array([0,0,0,0.5480])
h0_tot = h0 + h_ext
h_tot = h1 * contrast + h0_tot

r_fp, _ = get_ss(fit_c, W, h1, h0_tot, contrast) #, z_lims=z_lims, N_z=N_z)
A, B, init_perturb = get_params(r_fp, W, TAU, tau_noise, sigma_noise, init_perturb_mag, dt, variable, noise_type)
x_trajectory, _ = sim_global(r_fp, init_perturb_mag, W, h_tot, TAU, variable, T_sim=T_sim, dt=dt)
fig = plot_trajectory(x_trajectory, r_fp, W, h_ext, h_tot, variable, contrast, k, T_sim=T_sim, dt=dt, sim_type='local', noise_type=noise_type, savefig=False)
```

```python
# curve fitting and hopf analysis

freq_arr, amp_arr = get_freq_amp_arrays(
        r_fp, params, contrast, TAU, T_sim, dt, h_ext_lower, h_ext_upper, points,
        init_perturb_mag, variable='rate', threshold_factor=0.001, min_peak_distance=10, start_skip_fraction=0.5
    )
fit_results = fit_amplitude_curves(
        freq_arr, amp_arr, h_ext_lower, h_ext_upper, points,
        amplitude_threshold=0.005, min_points_after_start=5
    )
fig, axes = plot_freq_amp_vs_hext(
        freq_arr, amp_arr, h_ext_lower, h_ext_upper, points,
        fit_results=fit_results
    )
```
